Este script (proporcionado por Gemini) me va arreglar los datos históricos que bajo
de Binance, recordar que estos datos (en archivo CSV) contienen muchas columnas; entonces el script va a seleccionar solamente las columnas "Date" y "Close" y va a renombrar "Close" como "price" y guarda el dataframe con el nombre "datos_historia.csv"

```python
import pandas as pd

# --- Configuración ---
# Reemplaza 'nombre_de_tu_archivo.csv' con el nombre real de tu archivo de entrada.
archivo_entrada = 'solusdt_2023_5min.csv'
archivo_salida = 'datos_historia.csv'

# Columnas que queremos seleccionar y el nuevo nombre para una de ellas.
columnas_a_usar = ['Date', 'Close']
mapa_renombre = {'Close': 'price'}

try:
    # --- Proceso ---

    # 1. Leer el archivo CSV, seleccionando únicamente las columnas que nos interesan.
    #    Esto es más eficiente en memoria si el archivo original es muy grande.
    print(f"Leyendo el archivo '{archivo_entrada}'...")
    df = pd.read_csv(archivo_entrada, usecols=columnas_a_usar)

    # 2. Renombrar la columna 'Close' a 'price'.
    #    El método `rename` devuelve un nuevo DataFrame con los nombres cambiados.
    df_renombrado = df.rename(columns=mapa_renombre)

    # 3. Guardar el DataFrame resultante en un nuevo archivo CSV.
    #    `index=False` evita que se guarde una columna de índice no deseada.
    print(f"Guardando los datos procesados en '{archivo_salida}'...")
    df_renombrado.to_csv(archivo_salida, index=False)

    print("\n¡Proceso completado exitosamente!")
    print(f"Se ha creado el archivo '{archivo_salida}' con las columnas 'Date' y 'price'.")

    # Opcional: Muestra las primeras 5 filas del nuevo DataFrame para verificar.
    print("\nVista previa de los datos guardados:")
    print(df_renombrado.head())

except FileNotFoundError:
    print(f"\nError: No se pudo encontrar el archivo '{archivo_entrada}'.")
    print("Por favor, asegúrate de que el nombre del archivo es correcto y está en la misma carpeta que el script.")
except KeyError as e:
    print(f"\nError: Una de las columnas especificadas no se encontró en el archivo: {e}")
    print("Por favor, verifica que las columnas 'Date' y 'Close' existan en tu CSV.")
```
